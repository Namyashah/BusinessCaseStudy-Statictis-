# Q1

### Create a random sample of 50 exam scores (between 40 and 100).
### Calculate the sample mean and standard deviation.
### Use scipy.stats.t.interval() to compute the 95% confidence interval.
### Increase the sample size to 500 and recalculate the confidence interval.
### Compare how the interval changes with a larger sample size.

#### Created dataset And Computed Mean And Standard Deviation.

```python
exam_score = np.random.randint(40,101,size=50)
mean = np.mean(exam_score)
standard_deviation = np.std(exam_score,ddof=1)
print(f"Mean : {mean}")
print(f"standard Deviation : {standard_deviation:.4f}")
```

### Calculating Confidence Interval at 95% .

```python
confidence = 0.95
n = len(exam_score)
sem = stats.sem(exam_score)

ci = stats.t.interval(confidence,n-1,loc=mean,scale=sem)
print(f"Sample Mean : {mean:.4f}")
print(f"95% Confidence Interval : {ci}")
```

### Now lets Calculate Again With Sample Size Of 500.

```python
exam_score1 = np.random.randint(40,101,size=500)
mean1 = np.mean(exam_score1)
standard_deviation1 = np.std(exam_score1,ddof=1)
print(f"Mean : {mean1}")
print(f"standard Deviation : {standard_deviation1:.4f}")
```

```python
confidence = 0.95
n = len(exam_score1)
sem1 = stats.sem(exam_score1)

ci1 = stats.t.interval(confidence,n-1,loc=mean1,scale=sem1)
print(f"Sample Mean : {mean1:.4f}")
print(f"95% Confidence Interval : {ci1}")
```

The larger The sample Size The Precise The Confidence Interval Become As The Standard Of Error Become Smaller.

# Q2

### Assume the population mean is 70 and the standard deviation is 10.
### Generate a random sample of size 100.
### Use the z-test formula manually or with statsmodels.stats.weightstats.ztest( to
### check if the sample mean is significantly different from the population mean.
### Change the sample mean slightly and re-run the z-test.
### Interpret the z-stat and p-value.

```python
sample_population = np.random.normal(loc=70,scale=10,size=100)
sample_mean = np.mean(sample_population)

standard_error = 10/np.sqrt(100)
mean_differences = sample_mean-70

z_score = mean_differences/standard_error
print(f"Population Mean : 70")
print(f"Sample Mean : {sample_mean}")
```

```python
p_value = 2 * (1-stats.norm.cdf(abs(z_score)))
print(f"Z-Score : {z_score:.4f}")
print(f"P-Value : {p_value:.4f}")
```
