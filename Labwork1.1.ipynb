# Q4

### Use Matplotlib or Seaborn to plot:
- Histogram for age distribution.
- Bar chart for department count.
- Pie chart of department percentages.
- Boxplot for salary.

Write 1-2 lines describing the pattern seen in each chart.

```python
import matplotlib.pyplot as plt

plt.hist(df["Age"], bins=10, color="purple", edgecolor="black")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution of Employees")
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```

```python
counts = df["Department"].value_counts()

plt.bar(counts.index,counts.values,color = "teal",edgecolor = "black")
plt.xlabel("Departments")
plt.ylabel("Count Of Departments")
plt.title("Department By Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(11,6))
plt.pie(counts.values,labels=counts.index,autopct="%d%%",startangle=0,)
plt.title("Department Wise Distribution")
plt.tight_layout()
plt.legend()
plt.show
```

```python
plt.boxplot(df["Salary"],notch=True,meanline=True,vert=False)
plt.title("Salary Wise Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show
```
