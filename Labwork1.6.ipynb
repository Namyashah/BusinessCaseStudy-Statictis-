```python
maths_score = np.random.randint(40,100,size=50)
```

```python
physics_score = np.random.randint(40,100,size=50)
```

```python
df = pd.DataFrame({
    "Maths" : maths_score,
    "Physics" : physics_score 
})
df.head()
```

Doing Correlation And  Using Built in Function.

### Covariance

```python
covariance = df.cov()
print(f"Covariance:\n{covariance}\n")
```

### Pearson Correlation

```python
pearson_correlation = df.corr(method="pearson")
print(f"Pearson Correlation:\n{pearson_correlation}\n")
```

```python
shuffled_physics_score = np.random.shuffle(physics_score)
spearmen_correlation = df.corr(method="spearman")
spearmen_correlation
```

| Feature                | Pearson              | Spearman                    |
| ---------------------- | -------------------- | --------------------------- |
| Measures               | Linear relationship  | Monotonic relationship      |
| Based on               | Actual values        | Ranked values               |
| Sensitive to shuffling | Highly               | Highly                      |
| Original data          | 0.21 (weak positive) | \~0.25 (if monotonic trend) |
| After shuffling        | Near 0               | Near 0                      |
