### Create a visual mind map or list to differentiate between the two types.

Descriptive Statistics : Descriptive statistics refers to techniques for summarizing or presenting data in a meaningful way.

Inferential Statistics : Inferential statistics is a branch of statistics that allows us to make predictions, conclusions, or decisions about a population based on data collected from a sample.

| Feature           | Descriptive Statistics               | Inferential Statistics                             |
| ----------------- | ------------------------------------ | -------------------------------------------------- |
| **Definition**    | Summarizes and describes data        | Makes predictions or inferences about a population |
| **Purpose**       | Organize, simplify, and present data | Generalize from sample to population               |
| **Example Tools** | Mean, Median, Charts, Tables         | Hypothesis Testing, Confidence Intervals           |

# Q3

### Calculate the mean of a numeric column using:
df['column'].mean()

numpy.mean()
### Find the median using:
df['column'].median()

numpy.median()
### Find the mode using:

df['column'].mode()

Create a bar chart comparing Mean, Median, and Mode.

## Mean

```python
mean_dataset = dataset["Height"].mean()
mean_dataset
```

```python
np.mean(dataset["Height"])
```

## Median

```python
median_dataset = dataset["Height"].median()
median_dataset
```

```python
np.median(dataset["Height"])
```

## Mode

```python
mode_dataset = dataset["Height"].mode()[0]
mode_dataset
```

```python
x_axis = ["Mean","Median","Mode"]
y_axis = [mean_dataset,median_dataset,mode_dataset]

plt.bar(x_axis,y_axis,color=["teal","purple","lightblue"],edgecolor="black")
plt.grid(axis="y",linestyle="--",alpha=0.7)
plt.xlabel("Comparison")
plt.ylabel("Values")
plt.tight_layout()
plt.show()
```
