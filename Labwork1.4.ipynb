# Q1

### Simulate a fair dice roll 1000 times using numpy.random.randint().

### Calculate the probability of each outcome using frequency count.

### Simulate a coin toss 500 times.

### Estimate the probability of getting heads and tails.

### Use collections.Counter or value _counts() to calculate experimental probability.

#### Dice Roll

```python
dice_roll = np.random.randint(1,7,size=1000)

frequency = pd.value_counts(dice_roll)
number_list = [1,2,3,4,5,6]
n = 0
for counts in frequency:
    print(f"Dice Number {number_list[n]} : {counts}")
    n = n+1
```

#### Probability of each outcome.

```python
n = 0
sum_of_outcomes = np.sum(frequency)
for i in frequency:
    print(f"Probability Of {number_list[n]} Is :  {i/sum_of_outcomes}")
    n = n+1
```

### Coin Roll

```python
coin = ["Heads","Tails"]
n = 0
coin_roll = np.random.choice(coin,size=1000)
frequency1 = pd.Series(coin_roll).value_counts()
sum_of_outcomes1 = np.sum(frequency1)

probabilities = frequency1 / sum_of_outcomes1

for outcome, prob in probabilities.items():
    print(f"Probability of {outcome}: {prob:.3f}")
```

# Q2

### Simulate two sets of data with slightly different means.
### Perform a t-test using scipy. stats.ttest _ind to get the p-value.
### Interpret p-value in markdown. Mention what it implies for the null hypothesis.

```python
import numpy as np
from scipy import stats

np.random.seed(42)
data1 = np.random.normal(loc=50, scale=5, size=100)
data2 = np.random.normal(loc=52, scale=5, size=100)

print("Mean_Data1:", np.mean(data1))
print("Mean_Data2:", np.mean(data2))
```

```python
t_static,p_value = stats.ttest_ind(data1,data2,)
print(f"T-Statictis : {t_static}")
print(f"P-Value : {p_value:.4f}")
```

### T-Test Results Interpretation:

Null Hypothesis(Ho) : The means of two data are same.

Alternative Hypothesis(H1) The means of two dataset Are different.

        P-Value : 0.0001

        This Is Less Than 0.05 (Common Significance Level).

        So,We Reject Null Hypothesis (H0).

### Conclusion:

There Is A Strong Evidence to Suggest That means of the Two groups are Significanlty Different.So We Reject The Null Hypothesis.

# Q3

### Create a sample dataset with scores before and after a training program.
### Use a paired t-test or an independent t-test to check for improvement.
### Calculate a confidence interval for the mean using scipy.stats.sem and stats.t.interval(),
### Set a significance level (a = 0.05) and determine if the null hypothesis should be rejected.

```python
n_participants = 20
before_score = np.random.normal(loc=60,scale=5,size=n_participants)

after_score = before_score + np.random.normal(loc=10,scale=3,size=n_participants)

data = pd.DataFrame({
    "Participant" : range(1,n_participants+1),
    "Before_Training" : before_score.round(2),
    "After_Training" : after_score.round(2)
})
```

### Paired t-Test Done

```python
t_stat,p_value = stats.ttest_rel(before_score,after_score)
print(f"T-Statictis : {t_stat}")
print(f"P-Value : {p_value:.12f}")
```

### Calculating Confidence Interval

```python
mean = np.mean(after_score)
sem_deviation = stats.sem(after_score)

confidence = 0.95
ci = stats.t.interval(confidence,df=len(after_score-1),loc=mean,scale=sem_deviation)

print(f"Sample Mean: {mean:.2f}")
print(f"95% Confidence Interval: {ci}")
```

```python
if p_value > 0.05 :
    print("We fail To Reject Null Hypothesis!")
else :
    print("We reject The Null Hypothesis!")
```

### Yes The Null Hypothesis Should be rejected As The p-Value lies in The rejectance Area Of 0.05.
